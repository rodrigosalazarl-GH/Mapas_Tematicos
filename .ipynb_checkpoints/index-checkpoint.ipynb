# Thematics: the DDM (Dot Density Map)



````python
theMapAndData.region.value_counts()
````

### Preprocessing: Filtering

````python
# then

regions = ['CENTRAL AMERICA AND THE CARIBBEAN	', 'SOUTH AMERICA', 'NORTH AMERICA']

selected = theMapAndData[
    theMapAndData.region.isin(regions)
]

selected.plot()
````



### Preprocessing: aggregating

This is a DOT UNIT proposal:

````python
justDots=selected.copy()
dot_value = 100000
justDots['num_dots'] = (justDots['mobiles'] / dot_value).astype(int)
````

### Preprocessing: distributing dots

And here, we get a location for the dots:

````python
# 2. Generate the random dot locations using .sample_points()
justDots = justDots.sample_points(size=justDots['num_dots']).explode(index_parts=True)
````

You get this:

````python
justDots
````

````python
# what is it?
type(justDots)
````

### Preprocessing: building GDF from GS

Notice the multi index (MI), one for a country and for a dot. As we are planing to send these points to R, we avoid MI:

````python
justDots.reset_index(drop=True, inplace=True)
selected_dots=gpd.GeoDataFrame(geometry=justDots)
selected_dots
````

Now you have a GDF:

````python
selected_dots.info()
````

Juts plot the polygons as the base, and the points as the layer on top!

````python
base=selected.plot(facecolor="white",#color of polygon fill
               edgecolor='grey') #color of border
selected_dots.plot(markersize=0.005, color='red',ax=base)
````

It is generally recommended that DDMs use an **equal-area projection**, which benefits density and limits area distortion.

````python
base=selected.to_crs(8857).plot(facecolor="white",
                              edgecolor='grey')
selected_dots.to_crs(8857).plot(markersize=0.005, color='red',ax=base)
````

Let's keep the continent countries with that crs:

````python
selected_8857=selected.to_crs(8857)
selected_dots_8857=selected_dots.to_crs(8857)
````

# Thematics: the PSM (Proportional Symbol Map)



### Preprocessing: GDF of centroids

1. A location for a symbol, generally the polygon centroid.

````python
selected_8857_centroids = selected_8857.copy()

# selected_8857_centroids will have a new geometry:
selected_8857_centroids['geometry'] = selected_8857_centroids['geometry'].centroid
````

2. A size of varying values (not constant as in DDMs). Let's use population again.

````python
selected_8857_centroids['size'] = selected_8857_centroids['mobiles'].apply(lambda x: x**0.5/100)

## or
# import numpy as np
# selected_8857_centroids['size'] = selected_8857_centroids['mobiles'].apply(lambda x: np.log10(x))
````

We got the basics, then:

````python
base=selected_8857.plot(facecolor="white",
                      edgecolor='grey')
# Plot the centroids on top
selected_8857_centroids.plot(
    ax=base,
    markersize=selected_8857_centroids['size'],
    color='grey'
)
````

````python
# replace the points
selected_8857_centroids["geometry"]=selected_8857.representative_point()
````

### Preprocessing: Computing new columns

PPMs are not thematic maps to see precise values, but to reveal clear differences on the variable distribution used. As, human eyes are very limited to detect area differences, we may need to use some other tactics to help uncover some patterns.

For example, let's confirm if we have outliers:

````python
boxplotInfo=selected_8857_centroids.boxplot(column='mobiles',return_type="dict")
````

The object **boxplotInfo** showed the boxplot, and we confirm there are outliers. We can recover them like this:

````python
outliers=boxplotInfo['fliers'][0].get_ydata()
## see
outliers
````

Then, these are the outlying countries:

````python
selected_8857_centroids[selected_8857_centroids.mobiles.isin(outliers)]
````

Knowing this information, we can create a column as outlier flag:

````python
selected_8857_centroids['mobiles_outlier']=selected_8857_centroids.mobiles.isin(outliers)*1

## see
selected_8857_centroids
````

We can use that like this:

````python
base = selected_8857.plot(color='white', edgecolor='black', figsize=(10,10))

# Define your color map
mapcolor = {1: 'black', 0: 'lightgrey'}

# Plot the centroids on top
selected_8857_centroids.plot(
    ax=base,
    markersize=selected_8857_centroids['size'],
    color=selected_8857_centroids['mobiles_outlier'].map(mapcolor)
)
````

Here, we do a good job for outliers, but it is difficult to interprete the other ones. We could use some redundancy, but more work is needed:
1. Create two maps of centroids:

````python
# Create explicit copies of the DataFrames
selected_8857_centroids_out = selected_8857_centroids[selected_8857_centroids.mobiles_outlier==1].copy()
selected_8857_centroids_no_out = selected_8857_centroids[selected_8857_centroids.mobiles_outlier==0].copy()
````

2. Compute sizes in each:

````python
selected_8857_centroids_out['size'] = selected_8857_centroids_out['mobiles'].apply(lambda x: x**0.5 / 50)
selected_8857_centroids_no_out['size'] = selected_8857_centroids_no_out['mobiles'].apply(lambda x: x**0.5 / 50)
````

Now, plot the three layers, using some color gradient for the non-outliers:

````python
base=selected_8857.plot(color='white', edgecolor='grey',figsize=(10,10))

selected_8857_centroids_no_out.plot(
    ax=base,
    markersize=selected_8857_centroids_no_out['size'],
    edgecolor='grey',
    c=selected_8857_centroids_no_out['size'],
    cmap='Blues_r'
)
selected_8857_centroids_out.plot(
    ax=base,
    markersize=selected_8857_centroids_out['size'],
    color='orange'
)
````

# Thematics: Choropleths





````python
selected_8857.plot(selected_8857.mobiles/selected_8857.sq_km)
````

Choropleths are great to represent normalized indicators such as:
* Densities
* Ratios
* Rates
* Averages

Population by area is a density indicator. This is a real number (not an integer), so most of the time the color gradient might  not be
that useful to reveal some pattern visually. Then, we often discretize (or bin) the indicator computed.

### Preprocessing: Discretizing

Now, we want to cut the variable. Run the next code to make sure you have tose packages:

````python
## do you have these installed?
! pip show numba mapclassify numpy
````

We will discretize this:

````python
selected_8857['mobiles_density']=selected_8857.population/selected_8857.sq_km
````

Let's explore the bining algorithms:

````python
import mapclassify
import numpy as np

np.random.seed(12345) # so we all get the same results!

# let's try 5 intervals
K=5
theVar=selected_8857['mobiles_density']
# same interval width, easy interpretation
ei5 = mapclassify.EqualInterval(theVar, k=K)
# same interval width based on standard deviation, easy - but not as the previous one, poor when high skewness
msd = mapclassify.StdMean(theVar)
# interval width varies, counts per interval are close, not easy to grasp, repeated values complicate cuts
q5=mapclassify.Quantiles(theVar,k=K)

# based on similarity, good for multimodal data
mb5 = mapclassify.MaximumBreaks(theVar, k=K)
# based on similarity, good for skewed data
ht = mapclassify.HeadTailBreaks(theVar) # no K needed
# based on similarity, optimizer
fj5 = mapclassify.FisherJenks(theVar, k=K)
# based on similarity, optimizer
jc5 = mapclassify.JenksCaspall(theVar, k=K)
# based on similarity, optimizer
mp5 = mapclassify.MaxP(theVar, k=K)
````

How can we select the right classification?
Let me use the the Absolute deviation around class median (ADCM) to make the comparisson:

````python
class5 = ei5,msd, q5,mb5,  ht, fj5, jc5, mp5
# Collect ADCM for each classifier
fits = np.array([ c.adcm for c in class5])
# Convert ADCM scores to a DataFrame
adcms = pd.DataFrame(fits)
# Add classifier names
adcms['classifier'] = [c.name for c in class5]
# Add column names to the ADCM
adcms.columns = ['ADCM', 'Classifier']
````

Now, plot the **adcms**:

````python
adcms.sort_values('ADCM').plot.barh(x='Classifier')
````

Let's save the best strategy:

````python
selected_8857['mobiles_density_FJ5'] = fj5.yb
````

````python
# there you are
selected_8857[['mobiles_density','mobiles_density_FJ5']].head(20)
````

We could create a copy of the bins to add labels:

````python
# renaming
newLabelsForLevels={0:"0_VeryLow", 1:"1_Low", 2:"2_Middle", 3:"3_High", 4:"4_VeryHigh"}

selected_8857['mobiles_density_FJ5_cat']=selected_8857.loc[:,'mobiles_density_FJ5'].replace(newLabelsForLevels)

# we have
selected_8857[['mobiles_density','mobiles_density_FJ5','mobiles_density_FJ5_cat']].head(20)
````

We are ready for a discrete choropleth:

````python
import matplotlib.pyplot as plt

f, ax = plt.subplots(1, figsize=(10, 10))
selected_8857.plot(column='mobiles_density_FJ5_cat', # variable to plot
                 cmap='viridis_r', # set of colors
                 categorical=True, # can be interpreted as category
                 edgecolor='grey', # border color
                 linewidth=0.3, # width of border
                 legend=True, # need a legend?
                 ax=ax
       )

ax.set_axis_off()
````
